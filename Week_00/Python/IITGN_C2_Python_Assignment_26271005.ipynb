# Question 1

Clean Training Records Before Dataset Preparation

Difficulty: Foundational | Marks: 10

## Case study

An ML team receives training records from several internal systems. Each record should contain a learner identifier and an engagement score, but scores arrive in different types and identifiers sometimes contain unnecessary spaces. Incomplete records may contain None. Before these records enter a training dataset, the team needs predictable cleaned data and a way to identify rejected records.

## Coding task

Required function or decorator

```
INTERFACE
clean_training_records(records)
```

- Accept learner_id only when it is a string that remains non-empty after trimming surrounding whitespace.
- Accept score as an integer, float, or string convertible to a finite float. Reject Booleans, complex numbers, None, and other unsupported types. Accept only scores from 0 through 100.
- Return accepted records with a trimmed learner_id, a float score, and an eligible Boolean that is True when score is at least 60.
- Return rejected record positions using zero-based indexing. Preserve accepted-record order, retain repeated valid records, and ignore extra fields.

## Sample input

```
PYTHON
records = [
    {"learner_id": " L01 ", "score": "82.5"},
    {"learner_id": "L02", "score": 60},
    {"learner_id": "L03", "score": None},
    {"learner_id": "   ", "score": 75},
    {"learner_id": "L04", "score": True},
    {"learner_id": "L05", "score": 48.0},
    {"learner_id": "L06", "score": 105},
]
clean_training_records(records)
```

## Expected output

```
EXPECTED RESULT
{
    "cleaned": [
        {"learner_id": "L01", "score": 82.5, "eligible": True},
        {"learner_id": "L02", "score": 60.0, "eligible": True},
        {"learner_id": "L05", "score": 48.0, "eligible": False},
    ],
    "rejected_indices": [2, 3, 4, 6],
}
```

## Constraints and edge cases

- The outer input is a list. Reject elements that are not dictionaries and records missing required keys.
- Strings such as "nan" and "inf" are invalid because their converted values are not finite.
- An empty input returns empty cleaned and rejected_indices lists.
- Input records must remain unchanged.

## Student deliverables

Working code, visible sample output, at least three additional meaningful test cases, and a brief explanation of your approach.

## Interview follow up

Why does checking only whether a score is an integer risk accepting Boolean values?

````python
def clean_training_records(records) :
  count =-1
  clean_record = []
  rejected = []
  for record in records:
    count+=1
    #condition to check record is the dictonary and contains necesarry key to preocess data
    if isinstance(record,dict) and "learner_id" in record and "score" in record :
      learner_id = record["learner_id"].strip();
      #condition to check learner id is not empty and score is not boolean so while type casting it does not converts to int/float for processing
      if len(learner_id)>0 and not isinstance(record["score"],bool):
        try:
          score = float(record["score"])
          #condition to check the score is the range as well as to avoid nan and inf values
          if score >= 0 and score<=100:
            new_record = dict()
            new_record["learner_id"] = learner_id
            new_record["score"] = score
            new_record["eligible"] = True if score>=60.0 else False
            clean_record.append(new_record)
          else:
            rejected.append(count)
        #catching exceptions to handle all values types that can be converted into float
        except TypeError:
          rejected.append(count)
      else:
        rejected.append(count)
    else:
      rejected.append(count)
  result= dict();
  result["cleaned"] = clean_record
  result["rejected_indices"] = rejected
  return result
````

# `clean_training_records` —  Brief Explanation

## Overview

The `clean_training_records()` function validates and cleans a list of training records. It separates valid records from invalid ones and returns the indices of rejected records.

## Approach

The function processes each record one by one:

1. **Track the record index**

   * Maintains a `count` variable to identify the original index of each record.

2. **Validate record structure**

   * Checks whether the record is a dictionary.
   * Ensures the required keys `learner_id` and `score` are present.

3. **Validate `learner_id`**

   * Removes leading and trailing whitespace using `.strip()`.
   * Rejects the record if the resulting learner ID is empty.

4. **Validate `score`**

   * Rejects boolean values because `bool` is a subclass of `int` in Python and could otherwise be converted to `0.0` or `1.0`.
   * Converts valid numeric values to `float`.
   * Rejects scores outside the range **0–100**.
   * Invalid values that cannot be converted to `float` are rejected.

5. **Create cleaned record**

   * Creates a new dictionary containing:

     * `learner_id`
     * `score`
     * `eligible`
   * A learner is considered eligible when the score is **60 or above**.

6. **Store rejected indices**

   * For every invalid record, its original index is added to `rejected_indices`.

7. **Return the result**

   * The function returns a dictionary containing:

     * `cleaned`: list of valid and normalized records
     * `rejected_indices`: indices of rejected records

# Question 2

Prepare Mixed Text Payloads for an NLP Pipeline

Difficulty: Foundational | Marks: 10

## Case study

A text-classification service receives messages from web forms and message queues. Web forms provide strings, while queue consumers sometimes provide bytes or bytearray objects. The NLP pipeline needs consistent tokens before further processing. Your function must standardise valid messages without silently accepting undecodable binary content.

## Coding task

Required function or decorator

```
INTERFACE
prepare_messages(payloads, min_length=3)
```

- Accept strings directly. Decode bytes and bytearray using UTF-8. Reject unsupported types and binary payloads that cannot be decoded.
- Lowercase accepted text, split on whitespace, and remove only the characters . , ! ? ; : from both ends of each token. Keep tokens of at least min_length characters, preserving repetition and order.
- Return one token list per accepted payload, including empty lists; rejected payload positions; and a sorted list of unique retained tokens across all accepted payloads.
- Use a list comprehension for at least one token transformation or filtering operation and a set comprehension to construct the vocabulary.

## Sample input

```
PYTHON
payloads = [
    "  AI, helps TEAMS! ",
    b"Build reliable agents.",
    bytearray(b"AI helps."),
    None,
    b"\xff",
]
prepare_messages(payloads, min_length=3)
```


## Expected output

```
EXPECTED RESULT

        ["helps", "teams"],

    "rejected_indices": [3, 4],

```

## Constraints and edge cases

- The outer input is a list. min_length must be a positive integer excluding Booleans; otherwise raise ValueError.
- Empty or whitespace-only text is accepted and produces an empty token list.
- Internal punctuation is unchanged: "agent-based" remains "agent-based".
- Do not modify supplied bytearray objects. Empty input produces three empty output lists.

## Student deliverables

Working code, visible sample output, at least three additional meaningful test cases, and a brief explanation of your approach.

## Interview follow up

What practical difference between bytes and bytearray matters when processing input without modifying it?

````python
def prepare_messages(payloads, min_length=3):
  result = dict()
  count =0
  token_lists = []
  vocubabulary = set();
  reject_index = []
  for var in payloads :
    #To reject any other messages with other data types other than string , byte and bytearray
    if isinstance(var, str) or isinstance(var, bytes) or isinstance(var, bytearray):
      msg =""
      #flag to detect error while decode in the bytes stream and reject messages
      decode_error_flag = False
      if isinstance(var, bytes) or isinstance(var, bytearray):
        try:
          msg = var.decode("utf-8")
        except UnicodeDecodeError:
          decode_error_flag = True
      else:
        msg = var
      #Building the token from non error decoded string messages
      if not decode_error_flag:
        # Walrus operator with combination of comprehension to remove the invalid characters at start and end of the string
        valid_token_list = [ cleaned  for tken in msg.lower().strip().split() if len(cleaned := tken.strip(".,!?;: ")) >= min_length ]
        vocubabulary.update(valid_token_list)
        token_lists.append(valid_token_list)
      else:
        reject_index.append(count)
    else:
      reject_index.append(count)
    count+=1
  result["token_lists"] = token_lists
  result["vocabulary"] = list(vocubabulary)
  result["rejected_indices"] = reject_index
  return result
````

# `prepare_messages` — Brief Explanation

## Overview

The `prepare_messages()` function prepares a collection of messages for text processing. It accepts `str`, `bytes`, and `bytearray` values, converts byte-based messages to UTF-8 strings, tokenizes the messages, removes punctuation from token boundaries, filters short tokens, builds a unique vocabulary, and records rejected message indices.

## Approach

1. **Track the payload index**
   - Uses `count` to maintain the original index of every payload.

2. **Validate the input type**
   - Accepts only `str`, `bytes`, and `bytearray`.
   - Other data types are rejected.

3. **Decode byte-based messages**
   - `bytes` and `bytearray` values are decoded using UTF-8.
   - If decoding raises `UnicodeDecodeError`, the payload is rejected.
   - Normal strings are used directly.

4. **Normalize and tokenize the message**
   - Converts the message to lowercase using `.lower()`.
   - Removes leading/trailing whitespace using `.strip()`.
   - Splits the message into tokens using `.split()`.

5. **Clean and filter tokens**
   - Removes punctuation characters `.,!?;: ` from the beginning and end of each token.
   - Keeps only tokens whose length is at least `min_length`.

6. **Build the vocabulary**
   - Adds valid tokens to a `set`.
   - A set automatically removes duplicate words.

7. **Store token lists**
   - The cleaned token list for each valid message is added to `token_lists`.

8. **Track rejected messages**
   - Invalid data types and invalid UTF-8 messages are added to `rejected_indices`.

9. **Return the result**
   - Returns `token_lists`, `vocabulary`, and `rejected_indices` in a dictionary.

## Result Structure

```python

    ],
    "vocabulary": [

    ],
    "rejected_indices": [3, 4]
}
```

## Walrus Operator

The token filtering uses:

```python

    cleaned
    for tken in msg.lower().strip().split()
    if len(cleaned := tken.strip(".,!?;: ")) >= min_length
```

The same `cleaned` value is then:

1. Checked using `len(cleaned) >= min_length`.
2. Added to the resulting list if the condition is true.

This avoids calling `strip()` twice for the same token.

For example, a message such as:

produces a vocabulary containing:

```python

```

rather than storing `"ai"` twice.

# Question 3

Summarise Classification Prediction Results

Difficulty: Foundational to intermediate | Marks: 10

## Case study

A support-ticket classifier assigns categories such as billing, technical, and account. After an evaluation run, an engineer needs overall accuracy, performance by actual category, and the IDs of mistakes that need inspection. Some prediction records are incomplete, so the report must distinguish rejected data from valid but incorrect predictions.

## Coding task

Required function or decorator

```
INTERFACE
summarise_predictions(results)
```

- A valid record is a dictionary containing non-empty string values for id, actual, and predicted. Trim surrounding whitespace and lowercase the two labels.
- Return the valid-record count, correct-prediction count, and accuracy rounded to four decimal places.
- Return a nested dictionary with total and correct counts for each actual label, misclassified IDs in input order, and rejected record positions.
- Repeated records count as separate evaluation observations. Ignore extra fields.

## Sample input

```
PYTHON
results = [
    {"id": "T1", "actual": "Billing", "predicted": " billing "},
    {"id": "T2", "actual": "Technical", "predicted": "billing"},
    {"id": "T3", "actual": "billing", "predicted": "billing"},
    {"id": "T4", "actual": "Account", "predicted": "ACCOUNT"},
    {"id": "T5", "actual": None, "predicted": "account"},
]
summarise_predictions(results)
```

## Expected output

```
EXPECTED RESULT
{
    "valid_count": 4,
    "correct_count": 3,
    "accuracy": 0.75,
    "by_actual": {
        "billing": {"total": 2, "correct": 2},
        "technical": {"total": 1, "correct": 0},
        "account": {"total": 1, "correct": 1},
    },
    "misclassified_ids": ["T2"],
    "rejected_indices": [4],
}
```

## Constraints and edge cases

- The outer input is a list. Reject non-dictionary elements, missing required keys, and blank required strings.
- With no valid records, accuracy is None and counts are zero. The category summary and misclassified-ID list are empty.
- Include categories with zero correct predictions. Rejected positions follow input order.
- Do not modify the original records.

## Student deliverables

Working code, visible sample output, at least three additional meaningful test cases, and a brief explanation of your approach.

## Interview follow up

Why is None more informative than 0.0 when accuracy cannot be calculated?

````python
def summarise_predictions(results):
  summary_prediction = dict()
  valid_count = 0
  correct_count = 0
  indx =0
  accuracy = None
  by_actual = dict()
  misclassified_ids = []
  rejected_indices = []
  for rslt in results:
    # Condtion for result to be dictionary and it should contain all keys with String datatype for consideration for valid record
    if(isinstance(rslt, dict) and "id" in rslt and "actual" in rslt and "predicted" in rslt and isinstance(rslt["id"],str) and isinstance(rslt["actual"], str) and isinstance(rslt["predicted"], str)):
      id = rslt["id"].strip().lower()
      actual = rslt["actual"].strip().lower()
      predicted = rslt["predicted"].strip().lower()
      #condition for to check the that only non empty strings are considered as valid record
      if(len(id) == 0 or len(actual)==0 or len(predicted)==0):
        rejected_indices.append(indx)
        continue
      valid_count+=1
      #Checking if actual tag contains in the tags dictory
      if actual not in by_actual:
        by_actual[actual] = {"total":0, "correct":0}
      by_actual[actual]["total"]+=1
      #condtion to check the actual and predicted are there or not
      if(actual == predicted):
        correct_count+=1
        by_actual[actual]["correct"]+=1
      else:
        misclassified_ids.append(rslt["id"])
    else:
      rejected_indices.append(indx)
    indx+=1
  summary_prediction["valid_count"] = valid_count
  summary_prediction["correct_count"] = correct_count
  if valid_count>0:
    accuracy = round(correct_count/valid_count,4)
  summary_prediction["accuracy"] = accuracy
  summary_prediction["by_actual"] = by_actual
  summary_prediction["misclassified_ids"] = misclassified_ids
  summary_prediction["rejected_indices"] = rejected_indices
  return summary_prediction
````

# `summarise_predictions()` Function - Brief Explanation

The `summarise_predictions()` function processes a list of prediction results and generates a summary containing:

* Number of valid records


The values of `id`, `actual`, and `predicted` must be strings.

## Processing

For each record, the function performs the following checks:

1. **Validate the record**

   * Checks that the record is a dictionary.
   * Checks that `id`, `actual`, and `predicted` keys exist.
   * Checks that all three values are strings.

2. **Normalize the values**

   * Removes leading/trailing spaces using `.strip()`.
   * Converts values to lowercase using `.lower()`.

   * Accuracy is calculated only when there is at least one valid record.

```python
accuracy = round(correct_count / valid_count, 4)
```

````python
def update_annotation_queue(queue, incoming, completed,  priority=None, batch_size=3):

    # Validate batch_size: non-negative int, and bool is rejected
    # (bool is a subclass of int, so it needs an explicit check).
    if (isinstance(batch_size, bool) or not isinstance(batch_size, int) or batch_size < 0):
        raise ValueError("batch_size must be a non-negative integer")

    # 1. Combine into a NEW list (copy first so `queue` is never touched).
    combined = list(queue)
    combined.extend(incoming)

    # 2. Keep only the first occurrence of each ID.
    seen = set()
    unique = []
    for doc_id in combined:
        if doc_id not in seen:
            seen.add(doc_id)
            unique.append(doc_id)

    # 3. Remove completed IDs.
    done = set(completed)
    updated = [doc_id for doc_id in unique if doc_id not in done]

    # 4. Move priority to the front, preserving everyone else's order.
    if priority is not None and priority in updated:
        updated.remove(priority)
        updated.insert(0, priority)

    # 5. Build outputs.
    next_batch = updated[:batch_size]
    last_item = updated[-1] if updated else None
    positions = [(i, updated[i - 1])  for i in range(1, len(updated) + 1)]

    result = dict()
    result["queue"] = updated
    result["next_batch"] = next_batch
    result["last_item"] = last_item
    result["positions"] = positions

    return result
````

## Problem

An ML team shares a baseline configuration across experiments. Engineers need to tweak selected parameters and attach metadata per run. ## Code

```python
from copy import deepcopy
from typing import Any, Dict, Optional


def build_experiment_config(
    base: Dict[str, Any],
    *,
    run_name: str,
    parameter_updates: Optional[Dict[str, Any]] = None,
    **metadata: Any,
) -> Dict[str, Any]:
    """Create an independent experiment config from a shared baseline.

    Raises:
        ValueError: run_name is not a non-empty string after trimming.
        TypeError: parameter_updates is neither a dict nor None.
        KeyError: parameter_updates contains an unknown parameter.
    """
    # 1. Validate inputs first (fail fast)
    if not isinstance(run_name, str) or not run_name.strip():
        raise ValueError("run_name must be a non-empty string")

    if parameter_updates is None:
        parameter_updates = {}
    elif not isinstance(parameter_updates, dict):
        raise TypeError("parameter_updates must be a dict or None")

    # 2. Independent copy, including nested dictionaries
    config: Dict[str, Any] = deepcopy(base)

    # 3. Reject unknown parameters before applying any update
    unknown = [k for k in parameter_updates if k not in config["parameters"]]
    if unknown:
        raise KeyError(f"Unknown parameter(s): {', '.join(map(repr, unknown))}")

    # 4. Apply updates and merge metadata (copied so caller's objects aren't shared)
    config["parameters"].update(deepcopy(parameter_updates))
    config["metadata"].update(deepcopy(metadata))

    # 5. Add trimmed run name
    config["run_name"] = run_name.strip()
    return config
```

## How it works

1. **Validate first.** `run_name` must be a non-empty string after trimming (`ValueError`), and `parameter_updates` must be a dict or `None` (`TypeError`). Failing early means no work is wasted on bad input.
2. **Deep copy the baseline.** `copy.deepcopy(base)` recursively duplicates every nested dict, so edits to the result never touch `base`, and edits to `base` never leak into the result. A shallow copy (`base.copy()`) would still share the nested `parameters` and `metadata` dicts, which was the original bug.
3. **Check all update keys before applying any.** Unknown keys raise `KeyError`, and because the check happens before the update, a failed call never leaves a half-updated config.
4. **Apply updates and merge metadata.** `dict.update()` adds new keys and overwrites existing ones, which gives the required "replace when keys match" behavior. Values are deep-copied too, so mutable objects passed by the caller (like lists) aren't shared.
5. **Add the trimmed `run_name`** at the top level and return the new config.

## Concepts used

| Concept | Where it appears |
|---|---|
| Mutable vs. immutable objects | Why assignment (`b = a`) shares nested dicts |
| Shallow vs. deep copy | `deepcopy` instead of `.copy()` |
| Keyword-only parameters | The bare `*` makes `run_name` and `parameter_updates` named-only |
| `**kwargs` | `**metadata` collects extra named arguments into a dict |
| Type hints | `Dict[str, Any]`, `Optional[...]` (documentation only, not enforced at runtime) |
| Exceptions | `ValueError` (bad value), `TypeError` (bad type), `KeyError` (unknown key) |
| List comprehension | Building the `unknown` key list |
| `dict.update()` | Merging parameters and metadata |
| `str.strip()` | Trimming `run_name` and detecting empty strings |

## Edge cases handled

- `parameter_updates=None` or `{}` means no parameter changes.
- Whitespace-only `run_name` raises `ValueError`.
- Non-dict `parameter_updates` (e.g. a list of tuples) raises `TypeError`.
- Unknown parameter names raise `KeyError`, and the baseline stays untouched.
- Mutating the returned nested dicts does not affect `base`, and vice versa.

## One-minute summary

The function deep-copies the baseline so nothing is shared, validates the inputs, checks that every parameter update targets an existing key, applies the updates, merges the extra keyword arguments into the metadata, adds the trimmed run name, and returns the new config. The original baseline is never modified.

````python
from copy import deepcopy
from typing import Any, Dict, Optional


def build_experiment_config(
    base: Dict[str, Any],
    *,
    run_name: str,
    parameter_updates: Optional[Dict[str, Any]] = None,
    **metadata: Any,
) -> Dict[str, Any]:
    """Create an independent experiment config from a shared baseline.

    - Deep-copies `base` so nested dicts are never shared.
    - Adds the trimmed `run_name` at the top level.
    - Applies `parameter_updates` only to keys already in base["parameters"].
    - Merges extra keyword arguments into the copied metadata dict.

    Raises:
        ValueError: if run_name is not a non-empty string after trimming.
        TypeError: if parameter_updates is neither a dict nor None.
        KeyError: if parameter_updates contains an unknown parameter.
    """
    # Validate inputs before doing any work
    if not isinstance(run_name, str) or not run_name.strip():
        raise ValueError("run_name must be a non-empty string")

    if parameter_updates is None:
        parameter_updates = {}
    elif not isinstance(parameter_updates, dict):
        raise TypeError("parameter_updates must be a dict or None")

    # Independent copy, including nested dictionaries
    config: Dict[str, Any] = deepcopy(base)

    # Reject unknown parameters (check all before applying any)
    unknown = [k for k in parameter_updates if k not in config["parameters"]]
    if unknown:
        raise KeyError(f"Unknown parameter(s): {', '.join(map(repr, unknown))}")

    # Deep-copy update values too so the caller's objects aren't shared
    config["parameters"].update(deepcopy(parameter_updates))

    # Merge metadata, replacing matching keys
    config["metadata"].update(deepcopy(metadata))

    config["run_name"] = run_name.strip()
    return config
````

**Returns** a dictionary:

| Key | Meaning |
|---|---|
| `valid_count` | Number of accepted scores |
| `mean` | Mean of valid scores, rounded to 2 decimals (`None` if none are valid) |
| `top_scores` | Up to `top_n` tuples `(batch_index, item_index, float_score)` |

## Code

```python
import math


def aggregate_scores(*batches, top_n=3):
    """
    Aggregate valid scores across any number of batches.

    Each batch must be a list or tuple. Valid entries are finite ints/floats
    in [0, 100], excluding booleans. Invalid entries are skipped and reported
    by (batch_index, item_index). Input batches are never modified.

    Returns a dict with:
        valid_count        - number of accepted scores
        mean               - mean rounded to 2 decimals (None if no valid scores)
        top_scores         - up to top_n (batch_index, item_index, float_score)
        rejected_positions - (batch_index, item_index) of every invalid entry
    """
    # Validate top_n (bool is a subclass of int, so exclude it explicitly)
    if isinstance(top_n, bool) or not isinstance(top_n, int) or top_n < 0:
        raise ValueError("top_n must be a non-negative integer (not a Boolean)")

    # Validate batch types before processing
    for batch in batches:
        if not isinstance(batch, (list, tuple)):
            raise TypeError("each batch must be a list or tuple")

    def is_valid(value):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            return False
        if isinstance(value, float) and not math.isfinite(value):
            return False
        return 0 <= value <= 100

    valid = []      # (batch_index, item_index, float_score)
    rejected = []   # (batch_index, item_index)

    for b_idx, batch in enumerate(batches):
        for i_idx, item in enumerate(batch):
            if is_valid(item):
                valid.append((b_idx, i_idx, float(item)))
            else:
                rejected.append((b_idx, i_idx))

    valid_count = len(valid)

    # Generator expression used in the aggregate calculation
    total = sum(score for _, _, score in valid)
    mean = round(total / valid_count, 2) if valid_count else None

    # Rank: descending score, then lower batch index, then lower item index
    ranked = sorted(valid, key=lambda t: (-t[2], t[0], t[1]))
    top_scores = ranked[:top_n]

    return {
        "valid_count": valid_count,
        "mean": mean,
        "top_scores": top_scores,
        "rejected_positions": rejected,
    }
```

---

## How it works

1. **Validate `top_n`**: `bool` is a subclass of `int` in Python, so Booleans are excluded explicitly.
2. **Validate batch types**: any batch that is not a list or tuple raises `TypeError` before any processing.
3. **Check each entry (`is_valid`)**: rejects Booleans and non-numbers first, then non-finite floats (`nan`, `inf`), then checks the 0 to 100 range.
4. **Collect results**: nested `enumerate` loops walk batches and items in order, so `rejected` is automatically in batch-then-item order. Valid scores are stored as `(batch_index, item_index, float(score))`.
5. **Calculate the mean**: `sum(score for _, _, score in valid)` is a *generator expression* with tuple unpacking. It adds only the score from each tuple. The mean is `None` when there are no valid scores.
6. **Rank**: `sorted(valid, key=lambda t: (-t[2], t[0], t[1]))` sorts by score descending (negated), then batch index ascending, then item index ascending. `ranked[:top_n]` takes the top entries (empty if `top_n` is 0).
7. **Return** the result dictionary. Input batches are only read, never modified.

## Concepts used

| Concept | Where |
|---|---|
| Variable-length arguments (`*batches`) | Function signature |
| Generator expression | `sum(score for _, _, score in valid)` |
| Tuple unpacking and `_` throwaway variable | `for _, _, score in valid` |
| `sorted()` with `key` and `lambda` | Ranking |
| Tuple comparison for multi-level tie-breaking | Sort key |
| Negation to reverse one sort criterion | `-t[2]` |

## Example

```python
batch_a = [80, 90, None]
batch_b = (90, 70, True)
batch_c = []

aggregate_scores(batch_a, batch_b, batch_c, top_n=3)
```

```python
{
    "valid_count": 4,
    "mean": 82.5,
    "top_scores": [(0, 1, 90.0), (1, 0, 90.0), (0, 0, 80.0)],
    "rejected_positions": [(0, 2), (1, 2)],
}
```

`None` at `(0, 2)` and `True` at `(1, 2)` are rejected. The two scores of `90.0` tie, so the lower batch index (`0`) ranks first.

## Edge cases

| Input | Result |
|---|---|
| No batches | `valid_count=0`, `mean=None`, empty lists |
| No valid scores | `mean=None`, empty `top_scores` |
| `top_n=0` | Empty `top_scores` |
| `"90"` (numeric string) | Rejected, not converted |
| `nan`, `inf`, `True`, `False` | Rejected |
| Non-list/tuple batch | `TypeError` |
| `top_n=-1`, `top_n=True`, `top_n=2.5` | `ValueError` |

````python
import math


def aggregate_scores(*batches, top_n=3):
    """
    Aggregate valid scores across any number of batches.

    Each batch must be a list or tuple. Valid entries are finite ints/floats
    in [0, 100], excluding booleans. Invalid entries are skipped and reported
    by (batch_index, item_index). Input batches are never modified.

    Returns a dict with:
        valid_count        - number of accepted scores
        mean               - mean rounded to 2 decimals (None if no valid scores)
        top_scores         - up to top_n (batch_index, item_index, float_score)
        rejected_positions - (batch_index, item_index) of every invalid entry
    """
    # Validate top_n (bool is a subclass of int, so exclude it explicitly)
    if isinstance(top_n, bool) or not isinstance(top_n, int) or top_n < 0:
        raise ValueError("top_n must be a non-negative integer (not a Boolean)")

    # Validate batch types before processing
    for batch in batches:
        if not isinstance(batch, (list, tuple)):
            raise TypeError("each batch must be a list or tuple")

    def is_valid(value):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            return False
        if isinstance(value, float) and not math.isfinite(value):
            return False
        return 0 <= value <= 100

    valid = []      # (batch_index, item_index, float_score)
    rejected = []   # (batch_index, item_index)

    for b_idx, batch in enumerate(batches):
        for i_idx, item in enumerate(batch):
            if is_valid(item):
                valid.append((b_idx, i_idx, float(item)))
            else:
                rejected.append((b_idx, i_idx))

    valid_count = len(valid)

    # Generator expression used in the aggregate calculation
    total = sum(score for _, _, score in valid)
    mean = round(total / valid_count, 2) if valid_count else None

    # Rank: descending score, then lower batch index, then lower item index
    ranked = sorted(valid, key=lambda t: (-t[2], t[0], t[1]))
    top_scores = ranked[:top_n]

    return {
        "valid_count": valid_count,
        "mean": mean,
        "top_scores": top_scores,
        "rejected_positions": rejected,
    }
````

# `summarise_tool_calls(records)` – Brief Explanation

## Purpose
Takes a list of tool-call log records and returns a summary: how many unique calls there were, per-tool statistics, which calls failed, and a ranking of tools by failures.

Each record is expected to have: `call_id`, `tool`, `status` (`"ok"` / `"error"`), `duration_ms`.

## Steps

1. **Deduplicate by `call_id` (last wins)**
   Loop through the records and store each in the `latest` dict keyed by `call_id`. A later record with the same ID overwrites the earlier one completely. Input records are only read, never modified.

2. **Group by tool**
   `calls_by_tool` maps each tool name to the list of its final (deduplicated) records, using `setdefault(...).append(...)`.

3. **Build per-tool report (`by_tool`)**
   A dictionary comprehension over tools (sorted alphabetically) computes:
   - `calls` – number of calls
   - `successes` – count with status `"ok"`
   - `failures` – count with status `"error"`
   - `total_duration_ms` – sum of durations

4. **Collect failed call IDs**
   Pick the `call_id`s whose final status is `"error"` and sort them lexicographically.

5. **Rank tools**
   Sort tool names by: failures (descending), then total calls (descending), then tool name (ascending). Negating the numbers gives descending order inside a single ascending sort.

## Return Value

```python
{
    "unique_calls": <int>,          # number of distinct call_ids
    "by_tool": {...},               # per-tool stats, tools sorted A→Z
    "failed_call_ids": [...],       # sorted list of failed IDs
    "tool_ranking": [...],          # tools ordered by failures, calls, name
}
```

## Key Points
- **Last record wins** per `call_id`, so retries or updates are handled correctly.
- **Deterministic output**: sorting at every stage gives a stable, repeatable result.
- **Single pass logic** is simple: dedupe → group → aggregate → rank.
- **Complexity**: roughly O(n + t log t + f log f), where n = records, t = tools, f = failed calls.

````python
def summarise_tool_calls(records):
    # 1. Keep only the last record per call_id.
    #    A later record replaces the earlier one entirely.
    #    Records are only read, never modified.
    latest = {}
    for record in records:
        latest[record["call_id"]] = record

    final_records = list(latest.values())

    # 2. Group the final records by tool.
    calls_by_tool = {}
    for record in final_records:
        calls_by_tool.setdefault(record["tool"], []).append(record)

    # 3. Build the per-tool report with a dictionary comprehension.
    by_tool = {
        tool: {
            "calls": len(calls),
            "successes": sum(1 for c in calls if c["status"] == "ok"),
            "failures": sum(1 for c in calls if c["status"] == "error"),
            "total_duration_ms": sum(c["duration_ms"] for c in calls),
        }
        for tool, calls in sorted(calls_by_tool.items())
    }

    # 4. Failed call IDs, sorted lexicographically.
    failed_call_ids = sorted(
        call_id for call_id, r in latest.items() if r["status"] == "error"
    )

    # 5. Rank tools: failures desc, total calls desc, tool name asc.
    tool_ranking = sorted(
        by_tool,
        key=lambda t: (-by_tool[t]["failures"], -by_tool[t]["calls"], t),
    )

    return {
        "unique_calls": len(latest),
        "by_tool": by_tool,
        "failed_call_ids": failed_call_ids,
        "tool_ranking": tool_ranking,
    }
````

# validate_text_batch: Brief Explanation

## Problem
Several NLP functions accept batches of text, and each had its own copy-pasted validation with different rules (one accepted blank strings, another crashed on tuples). The goal is a single reusable decorator that enforces one consistent input contract.

## Solution Overview
`validate_text_batch` is a decorator that checks the `texts` argument **before** the wrapped function runs. If the input is invalid, an exception is raised and the wrapped function is never executed.

## How It Works

1. **Locate `texts`**: taken from the first positional argument (`args[0]`) or from `kwargs["texts"]`. If neither is supplied, a `TypeError` is raised.
2. **Check the container**: `texts` must be a `list` (`isinstance(texts, list)`). Tuples, sets, and plain strings raise `TypeError`.
3. **Pass 1, type check**: every item must be a `str`, otherwise `TypeError`.
4. **Pass 2, blank check**: any item that is empty after `.strip()` raises `ValueError`.
5. **Run the function**: only if all checks pass is `func(*args, **kwargs)` called, with all arguments forwarded unchanged.

## Why Two Passes?
The requirement says a batch containing both a non-string item and a blank string must raise `TypeError`. Checking all types first, then blanks, guarantees that priority regardless of item order.

## `text_lengths`
```python
@validate_text_batch
def text_lengths(texts: List[str], *, strip: bool = True) -> List[int]:
    return [len(t.strip()) if strip else len(t) for t in texts]
```
- `strip=True`: returns lengths after trimming surrounding whitespace.
- `strip=False`: returns lengths of the original strings.
- Internal whitespace always counts as characters.
- An empty list is valid and returns `[]`.

## Example
```python
texts = ["  agent ", "ML", "data"]
text_lengths(texts)                    # [5, 2, 4]
text_lengths(texts=texts, strip=False) # [8, 2, 4]
```

## Key Design Points
| Point | Detail |
|---|---|
| Runtime enforcement | Real `isinstance` checks, not type hints alone |
| Metadata preserved | `functools.wraps` keeps `__name__` and `__doc__` |
| No mutation | The input list is only read, never modified or copied |
| Exception types | `TypeError` for wrong types, `ValueError` for blank strings |
| Fail fast | Invalid input never reaches the wrapped function |

````python
from functools import wraps
from typing import Any, Callable, List, TypeVar

F = TypeVar("F", bound=Callable[..., Any])


def validate_text_batch(func: F) -> F:
    """Validate that `texts` is a list of non-blank strings before calling func."""

    @wraps(func)
    def wrapper(*args: Any, **kwargs: Any) -> Any:
        if args:
            texts = args[0]
        elif "texts" in kwargs:
            texts = kwargs["texts"]
        else:
            raise TypeError("texts must be provided as a list of strings")

        if not isinstance(texts, list):
            raise TypeError("texts must be a list")

        # Pass 1: type check the whole batch (TypeError takes priority).
        for item in texts:
            if not isinstance(item, str):
                raise TypeError("every item in texts must be a string")

        # Pass 2: blank check, only once all items are known to be strings.
        for item in texts:
            if not item.strip():
                raise ValueError("texts must not contain blank strings")

        return func(*args, **kwargs)

    return wrapper  # type: ignore[return-value]


@validate_text_batch
def text_lengths(texts: List[str], *, strip: bool = True) -> List[int]:
    """Return the character length of each text, optionally after trimming."""
    return [len(t.strip()) if strip else len(t) for t in texts]
````

# `track_calls` – Code Explanation

## Purpose
`track_calls(history)` is a decorator factory that builds an audit trail of every attempted call to a decorated function. Engineers can inspect `history` after a failed run to see which operations were attempted, which succeeded, and which raised exceptions.

## How it works
1. **Factory layer** – `track_calls(history)` captures the shared `history` list in a closure and returns a decorator.
2. **Decorator layer** – `decorator(func)` wraps the target function and returns `wrapper`.
3. **Wrapper layer** – on each call it:
   - computes `call_number = len(history) + 1`;
   - runs the original function with `*args, **kwargs`;
   - appends one dict with `call_number`, `function`, `status`, `error_type`;
   - returns the result unchanged, or re-raises the original exception.

## Design decisions
| Requirement | How it is met |
|---|---|
| Shared numbering across functions | Number is derived from `len(history)`, so any functions using the same list share one counter; separate lists stay independent |
| Failed calls consume a number | An entry is appended in the `except` block too |
| Return values untouched | `return result` is passed straight through; `None` still counts as success |
| Failures never hidden | Only `Exception` is caught, and a bare `raise` preserves the original exception and traceback |
| Any signature supported | `*args, **kwargs` forwards positional and keyword-only arguments |
| Name and docstring preserved | `functools.wraps(func)` |
| Original function name in log | `func.__name__` |

## Utility functions
- **`average_score(scores)`** – raises `ValueError` for an empty list, otherwise returns `sum(scores) / len(scores)` (a float, e.g. `70.0`).
- **`format_run_name(prefix, *, number)`** – returns `f"{prefix}_{number:02d}"`, zero-padding to at least two digits (`3` → `eval_03`, `123` → `eval_123`). `number` is keyword-only.

## Sample run
```python
history = []
average_score([60, 80])               # 70.0
format_run_name("eval", number=3)     # "eval_03"
average_score([])                     # raises ValueError
```

Resulting `history`:
```python
[
    {"call_number": 1, "function": "average_score",   "status": "success", "error_type": None},
    {"call_number": 2, "function": "format_run_name", "status": "success", "error_type": None},
    {"call_number": 3, "function": "average_score",   "status": "error",   "error_type": "ValueError"},
]
```

## Complexity
Each call adds O(1) work (one `len` and one `append`) on top of the wrapped function.

## Out of scope
Concurrency, recursion and nested decorated calls (numbering assumes sequential calls).

````python
from functools import wraps


def track_calls(history):
    """Decorator factory: record one audit entry per attempted call in `history`."""

    def decorator(func):
        @wraps(func)  # preserves __name__ and __doc__
        def wrapper(*args, **kwargs):
            # Numbering is derived from the shared list, so decorated
            # functions sharing a history share call numbers.
            call_number = len(history) + 1
            try:
                result = func(*args, **kwargs)
            except Exception as exc:
                history.append({
                    "call_number": call_number,
                    "function": func.__name__,
                    "status": "error",
                    "error_type": type(exc).__name__,
                })
                raise  # re-raise the original exception unchanged
            history.append({
                "call_number": call_number,
                "function": func.__name__,
                "status": "success",
                "error_type": None,
            })
            return result  # returned unchanged (None still counts as success)

        return wrapper

    return decorator


history = []


@track_calls(history)
def average_score(scores):
    """Return the mean of a non-empty list of scores."""
    if not scores:
        raise ValueError("scores must not be empty")
    return sum(scores) / len(scores)


@track_calls(history)
def format_run_name(prefix, *, number):
    """Return '<prefix>_<number>' with the number zero-padded to at least two digits."""
    return f"{prefix}_{number:02d}"
````
